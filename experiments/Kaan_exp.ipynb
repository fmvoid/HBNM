```python

# Load Kaan SC
import scipy.io
import matplotlib.pyplot as plt

kaan_mat_path = "/home/frank/HBNM/data/simulated_wc_data/DifMatrix.mat"
kaan_data = scipy.io.loadmat(kaan_mat_path)

# Extract all matrices into separate variables
W_1 = kaan_data['W_1']
W_2 = kaan_data['W_2']
W_3 = kaan_data['W_3']

# Plot each matrix
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
im0 = axs[0].imshow(W_1, aspect='auto', cmap='viridis')
axs[0].set_title('W_1')
plt.colorbar(im0, ax=axs[0], fraction=0.046, pad=0.04)

im1 = axs[1].imshow(W_2, aspect='auto', cmap='viridis')
axs[1].set_title('W_2')
plt.colorbar(im1, ax=axs[1], fraction=0.046, pad=0.04)

im2 = axs[2].imshow(W_3, aspect='auto', cmap='viridis')
axs[2].set_title('W_3')
plt.colorbar(im2, ax=axs[2], fraction=0.046, pad=0.04)

plt.tight_layout()
plt.show()
```
