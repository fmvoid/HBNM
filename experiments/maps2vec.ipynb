A Notebook to convert .pscalar.nii files into vectors

Outputs:
- Raw heterogeneity vectors
- Linearized heterogeneity vectors

```python
# Imports
import h5py
from hbnm.model.utils import linearize_map, normalize_sc
import numpy as np
import os
```

### SC

```python
# Find the SC

with h5py.File('/home/frank/HBNM/data/demirtas_neuron_2019.hdf5', 'r') as f:
    print(f.keys())
    sc = f['sc'][:]

sc = normalize_sc(sc[:180,:180])
```

```python
# Create output directory if it doesn't exist
os.makedirs('/home/frank/HBNM/data', exist_ok=True)

# Save normalized SC matrix
np.save('/home/frank/HBNM/data/SC_Left_Hemi.npy', sc)
```

### Raw maps

```python
# Load Myelin Map

# Load the myelin data
with h5py.File('/home/frank/HBNM/data/demirtas_neuron_2019.hdf5', 'r') as f:
    t1t2 = f['t1wt2w'][:]

# linearize the myelin data and get only L hemisphere
myelin_raw = t1t2[:180].reshape(1, 180)
```

```python
import nibabel as nib

# Load each pscalar.nii file and extract data as numpy arrays
dopamine_raw = nib.load('/home/frank/HBNM/data/maps/dopamine_avg.pscalar.nii').get_fdata()
dopamine_d1like_raw = nib.load('/home/frank/HBNM/data/maps/dopamine_d1like_avg.pscalar.nii').get_fdata()
dopamine_d2like_raw = nib.load('/home/frank/HBNM/data/maps/dopamine_d2like_avg.pscalar.nii').get_fdata()
gaba_a_raw = nib.load('/home/frank/HBNM/data/maps/gaba_a_avg.pscalar.nii').get_fdata()
muscarinic_raw = nib.load('/home/frank/HBNM/data/maps/muscarinic_avg.pscalar.nii').get_fdata()
nicotinic_raw = nib.load('/home/frank/HBNM/data/maps/nicotinic_avg.pscalar.nii').get_fdata()
NMDA_raw = nib.load('/home/frank/HBNM/data/maps/NMDA_avg.pscalar.nii').get_fdata()
norepinephrine_raw = nib.load('/home/frank/HBNM/data/maps/norepinephrine_avg.pscalar.nii').get_fdata()
serotonin_raw = nib.load('/home/frank/HBNM/data/maps/serotonin_avg.pscalar.nii').get_fdata()
```

```python
# Create the directory if it doesn't exist
output_dir = '/home/frank/HBNM/data/heterogeneity_vectors/raw'
os.makedirs(output_dir, exist_ok=True)

# Dictionary of all _raw variables to save
raw_variables = {
    'myelin': myelin_raw,
    'dopamine': dopamine_raw,
    'dopamine_d1like': dopamine_d1like_raw,
    'dopamine_d2like': dopamine_d2like_raw,
    'gaba_a': gaba_a_raw,
    'muscarinic': muscarinic_raw,
    'nicotinic': nicotinic_raw,
    'NMDA': NMDA_raw,
    'norepinephrine': norepinephrine_raw,
    'serotonin': serotonin_raw
}

# Save each variable as a .npy file
for name, data in raw_variables.items():
    filepath = os.path.join(output_dir, f'{name}_raw.npy')
    np.save(filepath, data)
    print(f'Saved {name}_raw.npy - shape: {data.shape}')

print(f'\nAll raw variables saved to: {output_dir}')
```

### Linearize all maps

```python
# Linearize all raw maps using the linearize_map function
print("Linearizing all raw maps...")

myelin_lin = linearize_map(myelin_raw)
dopamine_lin = linearize_map(dopamine_raw)
dopamine_d1like_lin = linearize_map(dopamine_d1like_raw)
dopamine_d2like_lin = linearize_map(dopamine_d2like_raw)
gaba_a_lin = linearize_map(gaba_a_raw)
muscarinic_lin = linearize_map(muscarinic_raw)
nicotinic_lin = linearize_map(nicotinic_raw)
NMDA_lin = linearize_map(NMDA_raw)
norepinephrine_lin = linearize_map(norepinephrine_raw)
serotonin_lin = linearize_map(serotonin_raw)

print("All maps linearized successfully!")
```

```python
# Create the linearized directory if it doesn't exist
linearized_dir = '/home/frank/HBNM/data/heterogeneity_vectors/linearized'
os.makedirs(linearized_dir, exist_ok=True)

# Dictionary of all linearized variables to save
linearized_variables = {
    'myelin': myelin_lin,
    'dopamine': dopamine_lin,
    'dopamine_d1like': dopamine_d1like_lin,
    'dopamine_d2like': dopamine_d2like_lin,
    'gaba_a': gaba_a_lin,
    'muscarinic': muscarinic_lin,
    'nicotinic': nicotinic_lin,
    'NMDA': NMDA_lin,
    'norepinephrine': norepinephrine_lin,
    'serotonin': serotonin_lin
}

# Save each linearized variable as a .npy file
for name, data in linearized_variables.items():
    filepath = os.path.join(linearized_dir, f'{name}_linearized.npy')
    np.save(filepath, data)
    print(f'Saved {name}_linearized.npy - shape: {data.shape}')

print(f'\nAll linearized variables saved to: {linearized_dir}')
```
